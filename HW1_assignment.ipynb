<font color='green'> For each demographic type separately -- age, gender, party, race, and education -- weight the poll by just that demographic type, in accordance with the population proportions given. Report the resulting poll results, and briefly (at most 3 sentences) describe what you observe.


For example, when weighted by race, you'll report:

Weighted by race --- Clinton: 0.531, Trump: 0.469

(your results might be different due to rounding/precision, but we do not expect the difference to be large)


```python
df_two_party = dfpoll[
    dfpoll["candidate"].isin(["Hillary Clinton", "Donald Trump"])
].copy()

# 1 = Clinton 
# 0 = Trump
df_two_party["Y"] = (
    df_two_party["candidate"] == "Hillary Clinton"
).astype(int)

demographic_types = ["age", "gender", "party", "race", "education"]

# ChatGPT assisted loop
for demographic_type in demographic_types:

    target = dfdemographic[
        (dfdemographic["Demographic_Type_1"] == demographic_type) &
        (dfdemographic["Demographic_Type_2"].isna())
    ].copy()

    target_groups = target["Demographic_1"].tolist()

    poll_subset = df_two_party[
        df_two_party[demographic_type].isin(target_groups)
    ].copy()

    weighted_clinton = 0

    for index, row in target.iterrows():

        group = row["Demographic_1"]
        population_share = row["Electoral_Proportion"]

        group_respondents = poll_subset[
            poll_subset[demographic_type] == group
        ]

        group_clinton_mean = group_respondents["Y"].mean()

        weighted_clinton = (
            weighted_clinton
            + population_share * group_clinton_mean
        )
# ---
    weighted_trump = 1 - weighted_clinton

    print(
        f"Weighted by {demographic_type} --- "
        f"Clinton: {weighted_clinton:.3f}, "
        f"Trump: {weighted_trump:.3f}"
    )
```

The weighted estimate changes based on the variable. Party weighting produced the largest shift which suggests that the differences between poll sample and population matter more for some variables than others. Weighting by age and race gives the same result while gender and educatiom have higher Clinton values.

### 3) Two-dimensional joint-distribution weighting

<font color='green'>Now, for each pair of demographic types in dfdemographic, do the same -- weight the poll by that pair of demographic types, in accordance with the given joint distributions, and briefly (at most 3 sentences) describe what you observe.

For example, when weighted by race and age, you'll find:

Weighted by age and race: Clinton: 0.523, Trump: 0.477

```python
df_two_party = dfpoll[
    dfpoll["candidate"].isin(["Hillary Clinton", "Donald Trump"])
].copy()

# calculate group means
df_two_party["Y"] = (
    df_two_party["candidate"] == "Hillary Clinton"
).astype(int)

# rows containing 2 variables
joint_targets = dfdemographic[
    dfdemographic["Demographic_Type_2"].notna()
].copy()

demographic_pairs = (
    joint_targets[
        ["Demographic_Type_1", "Demographic_Type_2"]
    ]
    .drop_duplicates()
)

# ChatGPT assisted code below: ---
for _, pair in demographic_pairs.iterrows():

    type_1 = pair["Demographic_Type_1"]
    type_2 = pair["Demographic_Type_2"]

    pair_targets = joint_targets[
        (joint_targets["Demographic_Type_1"] == type_1) &
        (joint_targets["Demographic_Type_2"] == type_2)
    ]

    weighted_clinton = 0

    for _, row in pair_targets.iterrows():

        group_1 = row["Demographic_1"]
        group_2 = row["Demographic_2"]
        population_share = row["Electoral_Proportion"]

        group_respondents = df_two_party[
            (df_two_party[type_1] == group_1) &
            (df_two_party[type_2] == group_2)
        ]

        group_mean = group_respondents["Y"].mean()

        weighted_clinton += population_share * group_mean
# ---

    weighted_trump = 1 - weighted_clinton

    print(
        f"Weighted by {type_1} and {type_2}: "
        f"Clinton: {weighted_clinton:.3f}, "
        f"Trump: {weighted_trump:.3f}"
    )
```

The pairs that include party are clustered close to 0.50 for Clinton and several pairs have higher Clifton shares. The largest is from weighting by gender and education while the lowest is from party and education. This might show that the choice of variable can affect the final polling estimate.

```python
# Used GPT to fix ordering error
pairs = [
    ("party", "gender"),
    ("race", "gender")
]
# ---

# ChatGPT assisted loop:
for type_1, type_2 in pairs:

    joint_data = dfdemographic[
        (dfdemographic["Demographic_Type_1"] == type_1) &
        (dfdemographic["Demographic_Type_2"] == type_2)
    ].copy()

    if len(joint_data) == 0:
        joint_data = dfdemographic[
            (dfdemographic["Demographic_Type_1"] == type_2) &
            (dfdemographic["Demographic_Type_2"] == type_1)
        ].copy()

        type_1, type_2 = type_2, type_1

    marginal_1 = dfdemographic[
        (dfdemographic["Demographic_Type_1"] == type_1) &
        (dfdemographic["Demographic_Type_2"].isna())
    ]

    marginal_2 = dfdemographic[
        (dfdemographic["Demographic_Type_1"] == type_2) &
        (dfdemographic["Demographic_Type_2"].isna())
    ]

    marginal_1_dict = dict(
        zip(
            marginal_1["Demographic_1"],
            marginal_1["Electoral_Proportion"]
        )
    )

    marginal_2_dict = dict(
        zip(
            marginal_2["Demographic_1"],
            marginal_2["Electoral_Proportion"]
        )
    )

    weighted_joint = 0
    weighted_independence = 0
    comparison_rows = []

    for _, row in joint_data.iterrows():

        group_1 = row["Demographic_1"]
        group_2 = row["Demographic_2"]

        supplied_joint = row["Electoral_Proportion"]

        independence_share = (
            marginal_1_dict[group_1]
            * marginal_2_dict[group_2]
        )

        group_respondents = df_two_party[
            (df_two_party[type_1] == group_1) &
            (df_two_party[type_2] == group_2)
        ]

        group_mean = group_respondents["Y"].mean()

        weighted_joint += supplied_joint * group_mean
        weighted_independence += independence_share * group_mean

        comparison_rows.append({
            type_1: group_1,
            type_2: group_2,
            "Joint_Share": supplied_joint,
            "Marginal_Product": independence_share,
            "Share_Difference": independence_share - supplied_joint,
            "Clinton_Mean": group_mean
        })

# ---
    comparison_table = pd.DataFrame(comparison_rows)

    comparison_table["Absolute_Difference"] = (
        comparison_table["Share_Difference"].abs()
    )

    print(f"\nWeighted by {type_1} and {type_2}")

    print(
        f"Supplied joint: Clinton = {weighted_joint:.3f}, "
        f"Trump = {1 - weighted_joint:.3f}"
    )

    print(
        f"Independence approximation: Clinton = {weighted_independence:.3f}, "
        f"Trump = {1 - weighted_independence:.3f}"
    )

    print("\nLargest cell differences:")
    print(
        comparison_table
        .sort_values("Absolute_Difference", ascending=False)
        .head(3)
    )
```

For party and gender, the independence approximation changes the weighted Clinton share slightly from 0.503 to 0.504. Similarly, for race and gender, the estimate changes only from 0.534 to 0.536. The largest cell-level differences happens in groups with different candidate preferences. These differences mostly offset each other. and the overall independence assumption shows very similar weighted results for both demographic pairs.

### Bonus 2 (up to 1 point): Raking

<font color='green'> Hint: at the heart of raking is calculating a weight for each survey respondent, so that the weights, when summed up, match the population on desired marginals as much as possible. 

For example, suppose we have a survey with two respondents, A likes chocolate and B hates chocolate. We know that in the population, 80\% of people like chocolate (the number is made up). So in a representative sample with 2 respondents, we would expect to see 1.6 respondents who like chocolate. Thus, one possible thing to do is we assign A a weight of 1.6 and B a weight of 0.4, and then use these weights to weigh their answers to the question we care about.

Of course, when the number of respondents and the number of demographic dimensions that we want to match get larger, finding the weights itself becomes harder. There are mainly two ways to do this. The first is through an iterative approach known as Iterative Proportional Fitting (IPF). IPF iterates through each demographic variable at a time, and adjusts the weights of all respondents through post stratification. IPF is easier to implement by hand by just following the instructions. The second approach is known as generalized raking, which is done through solving an optimization problem, and proposed by [Deville, J. C., Särndal, C. E., & Sautory, O. (1993)](https://www.tandfonline.com/doi/abs/10.1080/01621459.1993.10476369). This approach attempts to solve one undesirable outcome of IPF that the resulting weights may be unbalanced, which leads to higher variance in the final estimate. [This blog post](https://dev.to/potloc/generalized-raking-for-survey-weighting-2d1d) gives a nice introduction to generalized raking, while providing a code snippet for implementing it in Python (you will have to make suitable changes, of course.)
